# 📊 SentinelForge Dataset Analysis
This notebook provides an interactive environment to explore and visualize the adversarial AI datasets used in the SentinelForge red-teaming lab.

```python
import os
import yaml
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt

# Setup paths
base_dir = os.path.abspath('..')
dataset_dir = os.path.join(base_dir, 'datasets', 'attacks')

print(f"Dataset Directory: {dataset_dir}")
```

### Load and Parse Datasets

```python
def load_yaml(filepath):
    with open(filepath, 'r') as f:
        return yaml.safe_load(f)

single_turn_files = [f for f in os.listdir(dataset_dir) if f.endswith('.yaml')]
multi_turn_dir = os.path.join(dataset_dir, 'multi_turn')
multi_turn_files = [f for f in os.listdir(multi_turn_dir) if f.endswith('.yaml')]

all_items = []

# Process Single Turn
for file in single_turn_files:
    data = load_yaml(os.path.join(dataset_dir, file))
    if data:
        for item in data:
            all_items.append({
                'type': 'single_turn',
                'name': item.get('name', 'Unknown'),
                'category': item.get('category', 'unknown'),
                'severity': item.get('severity', 'unknown').upper(),
                'owasp_category': item.get('owasp_category', 'unknown')
            })

# Process Multi Turn
for file in multi_turn_files:
    data = load_yaml(os.path.join(multi_turn_dir, file))
    if data:
        all_items.append({
            'type': 'multi_turn',
            'name': data.get('name', 'Unknown'),
            'category': data.get('category', 'unknown'),
            'severity': data.get('severity', 'unknown').upper(),
            'owasp_category': 'N/A: Multi-turn' # Multi-turn yaml schema doesn't have owasp_category directly on root
        })

df = pd.DataFrame(all_items)
df.head()
```

### ⚠️ Severities Distribution

```python
severity_counts = df['severity'].value_counts()
print(severity_counts)

severity_counts.plot(kind='bar', color=['red', 'orange', 'gold', 'lightblue'])
plt.title('Attack Severities Distribution')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

### 🏷️ Attack Categories

```python
category_counts = df['category'].value_counts()
print(category_counts)

category_counts.plot(kind='barh', color='purple')
plt.title('Attack Categories')
plt.xlabel('Count')
plt.show()
```

### 🛡️ OWASP Classifications

```python
owasp_counts = df['owasp_category'].value_counts()
print(owasp_counts)

owasp_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, figsize=(8, 8))
plt.title('OWASP Top 10 Classifications')
plt.ylabel('')
plt.show()
```
